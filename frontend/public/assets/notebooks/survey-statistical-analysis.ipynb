# 3. CONFIDENCE INTERVALS

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("Database_Tubes.csv")

# A. One-Sample Confidence Interval for Proportion
# One-Sample CI - ChatGPT
p_chatgpt = df['AI_Used_ChatGPT'].mean()
n_chatgpt = df['AI_Used_ChatGPT'].count()
z = stats.norm.ppf(0.975)  # 95% confidence
margin_chatgpt = z * np.sqrt(p_chatgpt * (1 - p_chatgpt) / n_chatgpt)
print("A. One-Sample Confidence Interval for Proportion")
print("   CI untuk proporsi pengguna ChatGPT:")
print(f"    Proporsi: {p_chatgpt:.3f}")
print(f"    CI: ({p_chatgpt - margin_chatgpt:.3f}, {p_chatgpt + margin_chatgpt:.3f})")

# One-Sample CI - DALL-E
p_dalle = df['AI_Used_DALL_E'].mean()
n_dalle = df['AI_Used_DALL_E'].count()
margin_dalle = z * np.sqrt(p_dalle * (1 - p_dalle) / n_dalle)
print("   CI untuk proporsi pengguna DALL-E:")
print(f"    Proporsi: {p_dalle:.3f}")
print(f"    CI: ({p_dalle - margin_dalle:.3f}, {p_dalle + margin_dalle:.3f})")


# B. Two-Sample CI for Difference in Proportions
# Two-Sample CI - ChatGPT: Laki-laki vs Perempuan
df_gender = df[df['Jenis_kelamin'].isin(['Laki-laki', 'Perempuan'])]
p_m = df_gender[df_gender['Jenis_kelamin'] == 'Laki-laki']['AI_Used_ChatGPT'].mean()
p_f = df_gender[df_gender['Jenis_kelamin'] == 'Perempuan']['AI_Used_ChatGPT'].mean()
n_m = df_gender[df_gender['Jenis_kelamin'] == 'Laki-laki']['AI_Used_ChatGPT'].count()
n_f = df_gender[df_gender['Jenis_kelamin'] == 'Perempuan']['AI_Used_ChatGPT'].count()
diff1 = p_m - p_f
se1 = np.sqrt((p_m * (1 - p_m)) / n_m + (p_f * (1 - p_f)) / n_f)
margin1 = z * se1
print("\nB. Two-Sample CI for Difference in Proportions")
print("   CI Selisih proporsi pengguna ChatGPT (Laki-laki - Perempuan):")
print(f"    Selisih: {diff1:.3f}")
print(f"    CI: ({diff1 - margin1:.3f}, {diff1 + margin1:.3f})")

# Two-Sample CI - Copilot: SMA vs S1
df_edu = df[df['Pendidikan_terakhir'].isin(['SMA', 'S1'])]
p_sma = df_edu[df_edu['Pendidikan_terakhir'] == 'SMA']['AI_Used_Copilot'].mean()
p_s1 = df_edu[df_edu['Pendidikan_terakhir'] == 'S1']['AI_Used_Copilot'].mean()
n_sma = df_edu[df_edu['Pendidikan_terakhir'] == 'SMA']['AI_Used_Copilot'].count()
n_s1 = df_edu[df_edu['Pendidikan_terakhir'] == 'S1']['AI_Used_Copilot'].count()
diff2 = p_sma - p_s1
se2 = np.sqrt((p_sma * (1 - p_sma)) / n_sma + (p_s1 * (1 - p_s1)) / n_s1)
margin2 = z * se2
print("   CI Selisih proporsi pengguna Copilot (SMA - S1):")
print(f"    Selisih: {diff2:.3f}")
print(f"    CI: ({diff2 - margin2:.3f}, {diff2 + margin2:.3f})")
```

# 4. HYPOTHESIS TESTING

```python
import pandas as pd
from statsmodels.stats.proportion import proportions_ztest, confint_proportions_2indep

# Load the dataframe if not already loaded (assuming 'df' is available from previous cells)
# df = pd.read_csv("Database_Tubes.csv")

alpha = 0.05 # Menggunakan alpha 5%

print("# A. One-Sample Z-Test for Proportions")

# Uji 1. Proporsi pengguna layanan AI (misal: ChatGPT)
# H0: Proporsi pengguna ChatGPT adalah 0.75 (<= 0.75)
# H1: Proporsi pengguna ChatGPT lebih dari 0.75 (> 0.75)
# Kita akan melakukan uji satu sisi ke kanan
count_chatgpt = df['AI_Used_ChatGPT'].sum()
n_chatgpt = df['AI_Used_ChatGPT'].count()
null_proportion_chatgpt = 0.75

z_stat_chatgpt, p_value_chatgpt = proportions_ztest(count_chatgpt, n_chatgpt, value=null_proportion_chatgpt, alternative='larger')

print("\nUji 1: Proporsi Pengguna ChatGPT")
print(f"  Jumlah pengguna ChatGPT: {count_chatgpt}")
print(f"  Total responden: {n_chatgpt}")
print(f"  Proporsi sampel: {count_chatgpt/n_chatgpt:.3f}")
print(f"  Hipotesis Nol (H0): Proporsi pengguna ChatGPT <= {null_proportion_chatgpt}")
print(f"  Hipotesis Alternatif (H1): Proporsi pengguna ChatGPT > {null_proportion_chatgpt}")
print(f"  Z-statistik: {z_stat_chatgpt:.3f}")
print(f"  P-value: {p_value_chatgpt:.4f}")

if p_value_chatgpt < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti bahwa proporsi pengguna ChatGPT lebih dari {null_proportion_chatgpt}.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti bahwa proporsi pengguna ChatGPT lebih dari {null_proportion_chatgpt}.")


# Uji 2. Proporsi responden yang mengkhawatirkan kebocoran data pribadi
# H0: Proporsi responden yang mengkhawatirkan kebocoran data pribadi adalah 0.60 (>= 0.60)
# H1: Proporsi responden yang mengkhawatirkan kebocoran data pribadi kurang dari 0.60 (< 0.60)
# Kita akan melakukan uji satu sisi ke kiri
count_data_concern = df['AI_Concern_Bocornya_atau_penyalahgunaan_data_pribadi_dan_priv'].sum()
n_data_concern = df['AI_Concern_Bocornya_atau_penyalahgunaan_data_pribadi_dan_priv'].count()
null_proportion_data_concern = 0.60

z_stat_data_concern, p_value_data_concern = proportions_ztest(count_data_concern, n_data_concern, value=null_proportion_data_concern, alternative='smaller')

print("\nUji 2: Proporsi Responden yang Mengkhawatirkan Kebocoran Data Pribadi")
print(f"  Jumlah responden khawatir data pribadi: {count_data_concern}")
print(f"  Total responden: {n_data_concern}")
print(f"  Proporsi sampel: {count_data_concern/n_data_concern:.3f}")
print(f"  Hipotesis Nol (H0): Proporsi responden khawatir data pribadi >= {null_proportion_data_concern}")
print(f"  Hipotesis Alternatif (H1): Proporsi responden khawatir data pribadi < {null_proportion_data_concern}")
print(f"  Z-statistik: {z_stat_data_concern:.3f}")
print(f"  P-value: {p_value_data_concern:.4f}")

if p_value_data_concern < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti bahwa proporsi responden yang mengkhawatirkan kebocoran data pribadi kurang dari {null_proportion_data_concern}.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti bahwa proporsi responden yang mengkhawatirkan kebocoran data pribadi kurang dari {null_proportion_data_concern}.")

print("\n\n# B. Two-Sample Z-Test for Differences in Proportions")

# Uji 1. Perbedaan proporsi pengguna ChatGPT antara Laki-laki dan Perempuan
# H0: Proporsi pengguna ChatGPT pada Laki-laki sama dengan Perempuan (p_laki - p_perempuan = 0)
# H1: Proporsi pengguna ChatGPT pada Laki-laki berbeda dengan Perempuan (p_laki - p_perempuan != 0)
# Kita akan melakukan uji dua sisi
count_male = df[df['Jenis_kelamin'] == 'Laki-laki']['AI_Used_ChatGPT'].sum()
n_male = df[df['Jenis_kelamin'] == 'Laki-laki']['AI_Used_ChatGPT'].count()
count_female = df[df['Jenis_kelamin'] == 'Perempuan']['AI_Used_ChatGPT'].sum()
n_female = df[df['Jenis_kelamin'] == 'Perempuan']['AI_Used_ChatGPT'].count()

z_stat_gender, p_value_gender = proportions_ztest([count_male, count_female], [n_male, n_female], alternative='two-sided')

print("\nUji 1: Perbedaan Proporsi Pengguna ChatGPT (Laki-laki vs Perempuan)")
print(f"  Jumlah pengguna ChatGPT (Laki-laki): {count_male}, Total (Laki-laki): {n_male}, Proporsi: {count_male/n_male:.3f}")
print(f"  Jumlah pengguna ChatGPT (Perempuan): {count_female}, Total (Perempuan): {n_female}, Proporsi: {count_female/n_female:.3f}")
print("  Hipotesis Nol (H0): Proporsi pengguna ChatGPT sama antara Laki-laki dan Perempuan")
print("  Hipotesis Alternatif (H1): Proporsi pengguna ChatGPT berbeda antara Laki-laki dan Perempuan")
print(f"  Z-statistik: {z_stat_gender:.3f}")
print(f"  P-value: {p_value_gender:.4f}")

if p_value_gender < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti signifikan bahwa proporsi pengguna ChatGPT berbeda antara Laki-laki dan Perempuan.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti signifikan bahwa proporsi pengguna ChatGPT berbeda antara Laki-laki dan Perempuan.")

# Uji 2. Perbedaan proporsi pengguna Copilot antara Pendidikan SMA dan S1
# H0: Proporsi pengguna Copilot pada SMA sama dengan S1 (p_sma - p_s1 = 0)
# H1: Proporsi pengguna Copilot pada SMA lebih besar dari S1 (p_sma - p_s1 > 0)
# Kita akan melakukan uji satu sisi ke kanan
count_sma = df[df['Pendidikan_terakhir'] == 'SMA']['AI_Used_Copilot'].sum()
n_sma = df[df['Pendidikan_terakhir'] == 'SMA']['AI_Used_Copilot'].count()
count_s1 = df[df['Pendidikan_terakhir'] == 'S1']['AI_Used_Copilot'].sum()
n_s1 = df[df['Pendidikan_terakhir'] == 'S1']['AI_Used_Copilot'].count()

z_stat_edu, p_value_edu = proportions_ztest([count_sma, count_s1], [n_sma, n_s1], alternative='larger')

print("\nUji 2: Perbedaan Proporsi Pengguna Copilot (SMA vs S1)")
print(f"  Jumlah pengguna Copilot (SMA): {count_sma}, Total (SMA): {n_sma}, Proporsi: {count_sma/n_sma:.3f}")
print(f"  Jumlah pengguna Copilot (S1): {count_s1}, Total (S1): {n_s1}, Proporsi: {count_s1/n_s1:.3f}")
print("  Hipotesis Nol (H0): Proporsi pengguna Copilot pada SMA <= S1")
print("  Hipotesis Alternatif (H1): Proporsi pengguna Copilot pada SMA > S1")
print(f"  Z-statistik: {z_stat_edu:.3f}")
print(f"  P-value: {p_value_edu:.4f}")

if p_value_edu < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti bahwa proporsi pengguna Copilot di SMA lebih besar dari di S1.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti bahwa proporsi pengguna Copilot di SMA lebih besar dari di S1.")
```

# 5. CHI-SQUARE TESTS

```python
import pandas as pd
from scipy.stats import chisquare, chi2_contingency


df = pd.read_csv("Database_Tubes.csv")

alpha = 0.05

print("\n# A. Goodness of Fit Test")

# Uji 1: Goodness of Fit Test untuk variabel Jenis_kelamin
# H0: Distribusi Jenis_kelamin adalah uniform (proporsi Laki-laki = proporsi Perempuan)
# H1: Distribusi Jenis_kelamin tidak uniform
gender_counts = df['Jenis_kelamin'].value_counts()
total_count = gender_counts.sum()
expected_counts = [total_count / len(gender_counts)] * len(gender_counts)

chi2_stat_gender_gof, p_value_gender_gof = chisquare(gender_counts, f_exp=expected_counts)

print("\nUji 1: Goodness of Fit Test untuk Jenis Kelamin")
print(f"  Observed Counts:\n{gender_counts}")
print(f"  Expected Counts (Uniform):\n{expected_counts}")
print(f"  Hipotesis Nol (H0): Distribusi Jenis_kelamin adalah uniform")
print(f"  Hipotesis Alternatif (H1): Distribusi Jenis_kelamin tidak uniform")
print(f"  Chi-square Statistik: {chi2_stat_gender_gof:.3f}")
print(f"  P-value: {p_value_gender_gof:.4f}")

if p_value_gender_gof < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti bahwa distribusi Jenis_kelamin tidak uniform.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti bahwa distribusi Jenis_kelamin tidak uniform.")


print("\n\n# B. Independence Test")

# Uji 1: Independence Test antara Jenis_kelamin dan Penggunaan ChatGPT
# H0: Penggunaan ChatGPT independen terhadap Jenis_kelamin
# H1: Penggunaan ChatGPT dependen terhadap Jenis_kelamin
contingency_table_gender_chatgpt = pd.crosstab(df['Jenis_kelamin'], df['AI_Used_ChatGPT'])

chi2_stat_gender_chatgpt, p_value_gender_chatgpt, dof_gender_chatgpt, expected_gender_chatgpt = chi2_contingency(contingency_table_gender_chatgpt)

print("\nUji 1: Independence Test antara Jenis Kelamin dan Penggunaan ChatGPT")
print("  Tabel Kontingensi:")
display(contingency_table_gender_chatgpt)
print(f"  Hipotesis Nol (H0): Penggunaan ChatGPT independen terhadap Jenis_kelamin")
print(f"  Hipotesis Alternatif (H1): Penggunaan ChatGPT dependen terhadap Jenis_kelamin")
print(f"  Chi-square Statistik: {chi2_stat_gender_chatgpt:.3f}")
print(f"  P-value: {p_value_gender_chatgpt:.4f}")
print(f"  Derajat Kebebasan (DOF): {dof_gender_chatgpt}")

if p_value_gender_chatgpt < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti signifikan bahwa penggunaan ChatGPT dependen terhadap Jenis_kelamin.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti signifikan bahwa penggunaan ChatGPT dependen terhadap Jenis_kelamin.")

print("\n\n# C. Homogeneity Test")

# Uji 1: Homogeneity Test proporsi penggunaan AI (misal: Grammarly) di Domisili Bandung dan Jakarta
# H0: Proporsi pengguna Grammarly sama antara Domisili Bandung dan Jakarta
# H1: Proporsi pengguna Grammarly berbeda antara Domisili Bandung dan Jakarta
df_domisili = df[df['Domisili_saat_ini'].isin(['Bandung', 'Jakarta'])]
contingency_table_domisili_grammarly = pd.crosstab(df_domisili['Domisili_saat_ini'], df_domisili['AI_Used_Grammarly'])

chi2_stat_domisili_grammarly, p_value_domisili_grammarly, dof_domisili_grammarly, expected_domisili_grammarly = chi2_contingency(contingency_table_domisili_grammarly)

print("\nUji 1: Homogeneity Test proporsi pengguna Grammarly (Bandung vs Jakarta)")
print("  Tabel Kontingensi:")
display(contingency_table_domisili_grammarly)
print(f"  Hipotesis Nol (H0): Proporsi pengguna Grammarly sama antara Domisili Bandung dan Jakarta")
print(f"  Hipotesis Alternatif (H1): Proporsi pengguna Grammarly berbeda antara Domisili Bandung dan Jakarta")
print(f"  Chi-square Statistik: {chi2_stat_domisili_grammarly:.3f}")
print(f"  P-value: {p_value_domisili_grammarly:.4f}")
print(f"  Derajat Kebebasan (DOF): {dof_domisili_grammarly}")

if p_value_domisili_grammarly < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti signifikan bahwa proporsi pengguna Grammarly berbeda antara Domisili Bandung dan Jakarta.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti signifikan bahwa proporsi pengguna Grammarly berbeda antara Domisili Bandung dan Jakarta.")
```

# 6. KOLMOGOROV-SMIRNOV TESTS

```python
import pandas as pd
from scipy.stats import kstest, norm
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Database_Tubes.csv")

alpha = 0.05

# Uji 1: One-Sample Kolmogorov-Smirnov Test for Normality of Age
# H0: Distribusi usia = distribusi normal
# H1: Distribusi usia ≠ distribusi normal

age = df['Usia__dalam_tahun']

# Perform the Kolmogorov-Smirnov test for normality
# We need to specify the parameters (mean and standard deviation) of the normal distribution
# We can estimate these from the sample data
mean_age = age.mean()
std_age = age.std(ddof=1) # Use ddof=1 for sample standard deviation

ks_stat_age, p_value_age = kstest(age, 'norm', args=(mean_age, std_age))

plt.figure(figsize=(10, 6))
sns.histplot(age, kde=True, bins=20, color='skyblue')
plt.title('Distribusi Usia')
plt.xlabel('Usia (dalam tahun)')
plt.ylabel('Frekuensi')
plt.show()

print("\nUji 1: One-Sample Kolmogorov-Smirnov Test untuk Normalitas Usia")
print(f"  Hipotesis Nol (H0): Distribusi usia = distribusi normal")
print(f"  Hipotesis Alternatif (H1): Distribusi usia ≠ distribusi normal")
print(f"  Estimasi Mean (dari sampel): {mean_age:.3f}")
print(f"  Estimasi Standard Deviation (dari sampel): {std_age:.3f}")
print(f"  KS Statistik: {ks_stat_age:.3f}")
print(f"  P-value: {p_value_age:.4f}")

if p_value_age < alpha:
    print(f"  Kesimpulan: Tolak H0 pada tingkat signifikansi {alpha}. Ada bukti bahwa distribusi usia tidak mengikuti distribusi normal.")
else:
    print(f"  Kesimpulan: Gagal menolak H0 pada tingkat signifikansi {alpha}. Tidak ada cukup bukti bahwa distribusi usia tidak mengikuti distribusi normal.")
```
